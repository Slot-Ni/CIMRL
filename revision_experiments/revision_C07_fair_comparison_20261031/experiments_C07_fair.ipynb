```python
from pathlib import Path
import os,sys,subprocess
import torch
ROOT=Path.cwd().resolve()
assert (ROOT/'fair_comparison.py').is_file()
SOURCE=ROOT.parent/'results/server_BarsRev'  # 服务器最新原结果目录
DATA_ROOT=ROOT.parent
CACHE_ROOT=ROOT.parent/'Hyper_Distance'
WORKSPACE=ROOT/'results/fair_BarsRev'
# 本机分析下载快照时可改：
# SOURCE=Path.home()/'Downloads/server_BarsRev_results.zip'
# DATA_ROOT=Path('D:/YX/HyperGraphMatrix')
# CACHE_ROOT=Path('D:/YX/Hyper_Distance')
def call(*args):
    cmd=[sys.executable,'-u',str(ROOT/'fair_comparison.py'),*map(str,args)]
    result=subprocess.run(cmd,cwd=ROOT,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True)
    print(result.stdout)
    result.check_returncode()
print(sys.executable, 'torch',torch.__version__)
```
